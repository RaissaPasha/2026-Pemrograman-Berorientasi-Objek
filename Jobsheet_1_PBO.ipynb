# Praktikum 01 Tipe dasar Data

```python
# Integer
angka_int = 10
print("angka_int =", angka_int, "->", type(angka_int))

#Float
angka_float = 3.14
print("angka_float =", angka_float, "->", type(angka_float))

# Complex
angka_complex = 2 + 3j
print("angka_complex =", angka_complex, "->", type(angka_complex))

# Boolean
is_active = True
print("is_active =", is_active, "->", type(is_active))

# String
teks = "Hello, Python!"
print("teks =", teks, "->", type(teks))

# List
daftar_angka = [1, 2, 3, 4, 5]
print("daftar_angka =", daftar_angka, "->", type(daftar_angka))

# Tuple
koordinat = (10, 20)
print("koordinat =", koordinat, "->", type(koordinat))

# Dictionary
data_mahasiswa = {
    "nama": "Andi",
    "nim": "A11.2002.12345",
    "jurusan": "Teknik Informatika"
}
print("data_mahasiswa =", data_mahasiswa, "->", type(data_mahasiswa))

# Set
himpunan_angka = {1, 2, 3, 2, 1}
print("himpunan_angka =", himpunan_angka, "->", type(himpunan_angka))

# Contoh penggunaan konversi tipe data
nilai_str = "100"
print("\nnilai_str =", nilai_str, "->", type(nilai_str))
nilai_int = int(nilai_str)
print("nilai_int =", nilai_int, "->", type(nilai_int))
```

# Praktikum 2 Variabel dan Operasi Dasar

```python
# Pendeklarasian Variabel
nama = "Budi"
umur = 20
tinggi = 170.5
is_student = True

print("Nama=", nama)
print("Umur=", umur)
print("Tinggi=", tinggi)
print("Student=", is_student)

# Operasi Aritmatika
a = 10
b = 3

penjumlahan = a + b
pengurangan = a - b
perkalian = a * b
pembagian = a / b
pembagian_bulat = a // b
modulus = a % b
pangkat = a ** b

print("\nOPERASI ARITMETIKA =")
print("a =", a, ", b =", b)
print("Penjumlahan =", penjumlahan)
print("Pengurangan =", pengurangan)
print("Perkalian =", perkalian)
print("Pembagian =", pembagian)
print("Pembagian Bulat =", pembagian_bulat)
print("Modulus =", modulus)
print("Pangkat =", pangkat)

# Operasi Perbandingan
lebih_besar = a > b
kurang_dari = a < b
sama_dengan = a == b
tidak_sama = a != b
lebih_besar_sama = a >= b
kurang_sama = a <= b

print("\nOPERASI PERBANDINGAN")
print("a > b =", lebih_besar)
print("a < b =", kurang_dari)
print("a == b =", sama_dengan)
print("a != b =", tidak_sama)
print("a >= b =", lebih_besar_sama)
print("a <= b =", kurang_sama)

# Operasi Logika
x = True
y = False

logika_and = x and y
logika_or = x or y
logika_not = not x

print("\nOPERASI LOGIKA")
print("x =", x, ", y =", y)
print("x and y =", logika_and)
print("x or y =", logika_or)
print("not x =", logika_not)

# Contoh penggunaan di dalam percabangan
if a > b and b > 0:
  print("\nKondisi terpenuhi: a lebih besar dari b dan b positif")
else:
  print("\nKondisi tidak terpenuhi atau b <= 0.")

```

# Praktikum 3 Percabangan

```python
# IF Sederhana
nilai = 85
print("Contoh IF Sederhana")
if nilai >= 80:
  print("Selamat! Anda lulus dengan nilai tinggi.\n")

# IF-ELSE
umur = 17
print("Contoh IF-ELSE")
if umur >= 18:
  print("Anda sudah cukup umur untuk membuat SIM.\n")
else:
  print("Anda belum cukup umur untuk membuat SIM.\n")

# IF-ELIF-ELSE
hari = "Rabu"
print("Contoh IF-ELIF-ELSE")
if hari == "Senin":
  print("Hari Senin - Saatnya kembali bekerja!")
elif hari == "Selasa":
  print("Hari Selasa - Jadwal Rapat Mingguan")
elif hari == "Rabu":
  print("Hari Rabu - Ada diskon di beberapa toko.\n")
else:
  print("Hari lainnya - Atur jadwalmu dengan baik.\n")


# IF Bersarang (Nested IF)
suhu = 35
print("Contoh IF Bersarang (Nested IF):")
if suhu > 30:
  print("Cuaca cukup panas.")
  if suhu > 40:
    print("Bahkan sangat terik! Disarankan banyak minum air.")
  else:
    print("Masih relatif normal, tapi tetap jaga kesehatan.\n")
else:
  print("Cuaca sepertinya cukup sejuk.\n")

# Menggabungkan Percabangan dengan Operasi Logika
nilai_teori = 75
nilai_praktek = 80
print("Contoh IF dengan Operasi Logika AND/OR:")
if nilai_teori >= 70 and nilai_praktek >= 70:
  print("Anda lulus karena nilai teori dan praktek memadai.\n")
elif nilai_teori < 70 and nilai_praktek < 70:
  print("Anda perlu meningkatkan nilai teori dan praktek.")
elif nilai_teori < 70:
  print("Anda perlu meningkatkan nilai teori.")
else:
  print("Anda perlu meningkatkan nilai praktek.")

# Penggunaan If Ternary (atau Conditional Expression)
angka = -5
print("Contoh IF Ternary (Conditional Expression):")
status = "Positif" if angka > 0 else "Negatif atau Nol"
print("Angka =", angka, "->", status)
```

# Praktikum 4 Perulangan

```python
# FOR Loop dengan range()
print("1) FOR loop dengan range()")
for i in range(5):
  print("Perulangan ke-", i)
print()

# 2. FOR Loop untuk mengiterasi List
print("2) FOR loop mengiterasi list")
buah = ["apel", "mangga", "jeruk", "pisang"]
for item in buah:
  print("Buah:", item)
print()

# WHILE Loop
print("3) WHILE loop sederhana")
count = 0
while count < 5:
  print("count =", count)
  count += 1
print()

# 4. BREAK pada Loop
print("4) BREAK di dalam loop")
for i in range(10):
  if i == 3:
    print("Loop dihentikan pada i =", i)
    break
  print("i =", i)
print()

# 5. CONTINUE pada Loop
print("5) CONTINUE di dalam loop")
for i in range(5):
  if i == 2:
    print("Lewati i =", i, "dengan continue")
    continue
  print("i =", i)
print()

# 6. NESTED Loop (Loop Bersarang)
print("6) NESTED loop")
for i in range(3):
  for j in range(2):
    print(f"i={i}, j={j}")
print()

# 7. Memanfaatkan ELSE pada Loop
print("7) ELSE pada loop for/while")
for x in range(3):
  print("x =", x)
else:
  print("Loop for telah selesai tanpa break.\n")

y = 0
while y < 3:
  print("y =", y)
  y += 1
else:
  print("Loop while telah selesai tanpa break.\n")

# 8. PASS sebagai placeholder
print("8) PASS (placeholder)")
for i in range(3):
  if i == 1:
    pass
  print("i =", i)

```

# Penugasan 2 Cek Bilangan Genap, Ganjil, Prima

```python
bilangan = int(input("Masukkan bilangan bulat : "))

if bilangan % 2 == 0:
  sifat_dasar = "Genap"
else:
  sifat_dasar = "Ganjil"

is_prima = True
if bilangan <= 1:
  is_prima = False
else:
  for i in range (2, int(bilangan**0.5) + 1):
    if bilangan % i == 0:
      is_prima = False
      break

status_prima = "dan merupakan bilangan Prima" if is_prima else "dan bukan bilangan Prima"
print(f"Bilangan {bilangan} adalah {sifat_dasar} {status_prima}.")
```
